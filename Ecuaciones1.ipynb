```python
import numpy as np

# Definimos la función
def f(x: float) -> float:
    return 1 / 4 * (x**3 + 3 * x**2 - 6 * x - 8)

# Importamos las librerías necesarias
import matplotlib.pyplot as plt

# Generamos los valores de x para la gráfica
x = np.linspace(-4, 2, 1000)

# Calculamos los valores de f(x) para la gráfica
ys = [f(xi) for xi in x]

# Método de bisección para encontrar la raíz
def biseccion(f, a, b, tol=1e-5, max_iter=100):
    if f(a) * f(b) >= 0:
        raise ValueError("El intervalo no contiene una raíz.")
    for _ in range(max_iter):
        c = (a + b) / 2
        if abs(f(c)) < tol or (b - a) / 2 < tol:
            return c
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    raise ValueError("El método no convergió.")

# Calculamos la raíz
# Adjust the interval to ensure it contains a root
raiz = biseccion(f, -2, 0)

# Generamos la gráfica
plt.plot(x, ys, label='f(x)')
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')  # Línea horizontal en y=0
plt.scatter(raiz, f(raiz), color='red', label=f'Raíz: x={raiz:.4f}')  # Punto de la raíz
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Método de Bisección - Raíz de f(x)')
plt.legend()
plt.grid(True)
plt.show()
```
